# ⌨️ Jupyter Notebook Shortcuts — VS Code

## 🚀 Running Cells

| Shortcut | Action |
|---|---|
| `Shift + Enter` | Run current cell and move to the next cell |
| `Ctrl + Enter` | Run current cell and stay on the same cell |
| `Alt + Enter` | Run current cell and insert a new cell below |

---

## ➕ Creating Cells

| Shortcut | Action |
|---|---|
| `Esc` → `A` | Add a cell **above** |
| `Esc` → `B` | Add a cell **below** |
| `Esc` → `D` → `D` | Delete current cell |

> These are Jupyter command-mode shortcuts. Press `Esc` first.

---

## 🔄 Changing Cell Type

| Shortcut | Action |
|---|---|
| `Esc` → `M` | Change cell to **Markdown** |
| `Esc` → `Y` | Change cell to **Code** |

---

## 🧭 Navigation

| Shortcut | Action |
|---|---|
| `Esc` | Enter command mode |
| `Enter` | Enter edit mode |
| `↑ / ↓` | Move between cells |
| `Ctrl + Enter` | Run without moving |

---

## 🛠️ Useful VS Code Shortcuts

| Shortcut | Action |
|---|---|
| `Ctrl + Shift + P` | Open Command Palette |
| `Ctrl + S` | Save notebook |
| `Ctrl + F` | Find text |
| `Ctrl + Z` | Undo |
| `Ctrl + Shift + Z` | Redo |

---

## ⭐ Most Important

For everyday ML work, remember these:

```text
Shift + Enter → Run cell
Ctrl + Enter  → Run and stay
Esc + A       → Add cell above
Esc + B       → Add cell below
Esc + M       → Markdown
Esc + Y       → Code
Esc + D + D   → Delete cell
```
